```python
# Definición de la proposición original
lloviendo = False

# Negación de la proposición
negacion = not lloviendo

# Imprimir los resultados
print(f"Proposición original, Está lloviendo: {lloviendo}")
print(f"Negación de la proposición original, Está lloviendo: {negacion}")
```

```python
paga_suscripcion = True
tiene_acceso = False

# La regla lógica: Si P entonces Q
regla_valida = not paga_suscripcion or tiene_acceso

print(f"¿La regla de negocio se cumple?: {regla_valida}")
```

```python
# Definición de dos proposiciones
ayuno = True
adelgazo = True

# Conjunción de las dos proposiciones
conjuncion = ayuno and adelgazo

# Imprimir los resultados
print(f"Proposición 1, Hago ayuno: {ayuno}")
print(f"Proposición 2, Adelgazo: {adelgazo}")
print(f"Conjunción de las proposiciones: {conjuncion}")
```
