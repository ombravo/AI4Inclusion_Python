```python
import numpy as np
import statistics as st
import matplotlib.pyplot as plt
from google.colab import output

# pesos de las zanahorias en gramos
zanahorias = [300, 280, 290, 310, 275, 290, 295, 315, 290, 280, 310, 305]

# Estadísticas descriptivas iniciales
media = np.mean(zanahorias)  # promedio
mediana = np.median(zanahorias)  # mediana, valor central
moda = st.mode(zanahorias)  # moda, el valor más repetido
desviacion = np.std(zanahorias)  # desviación estándar

# resultados iniciales
print("Análisis del concurso de zanahorias 1.0 🥕🏆")
print(f"Media inicial: {media} gramos")
print(f"Mediana inicial: {mediana} gramos")
print(f"Moda inicial: {moda} gramos")
print(f"Desviación estándar inicial: {desviacion:.1f} gramos\n")

# Histograma de la distribución de pesos
plt.hist(zanahorias, bins=6, color='orange', edgecolor='black')
plt.axvline(media, color='red', linestyle='dashed', linewidth=2, label=f'Media: {media:.1f}g')
plt.title('Distribución de pesos de las zanahorias concursantes')
plt.xlabel('Peso (gramos)')
plt.ylabel('Cantidad de zanahorias')
plt.legend()

# Forzar el renderizado y despliegue del gráfico antes de pedir los datos
plt.show()

# Solicitar datos con un diálogo prompt nativo de Javascript en Colab
print("Esperando entrada de datos...")
entrada_js = output.eval_js('prompt("Ingresa los pesos de las nuevas zanahorias separados por comas (ej: 320, 340, 285):")')

if entrada_js:
    try:
        nuevas_zanahorias = [float(peso.strip()) for peso in entrada_js.split(",") if peso.strip()]
        todas_zanahorias = zanahorias + nuevas_zanahorias

        media_acumulada = np.mean(todas_zanahorias)
        mediana_acumulada = np.median(todas_zanahorias)
        try:
            moda_acumulada = st.mode(todas_zanahorias)
        except st.StatisticsError:
            moda_acumulada = "No única"
        desviacion_acumulada = np.std(todas_zanahorias)

        print(f"\n--- Análisis de todos los datos combinados ({len(todas_zanahorias)} zanahorias) ---")
        print(f"Nueva Media: {media_acumulada:.1f} gramos")
        print(f"Nueva Mediana: {mediana_acumulada:.1f} gramos")
        print(f"Nueva Moda: {moda_acumulada} gramos")
        print(f"Nueva Desviación Estándar: {desviacion_acumulada:.1f} gramos")
    except ValueError:
        print("\nError: Asegúrate de ingresar únicamente números separados por comas.")
else:
    print("\nNo se ingresaron nuevos datos. Se conservan las estadísticas iniciales.")
```

```python
import numpy as np
import statistics as st
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. DATOS DE ENTRADA
# ---------------------------------------------------------
# Pesos de las zanahorias en gramos
zanahorias = [300, 280, 290, 310, 275, 290, 295, 315, 290, 280, 310, 305]

# ---------------------------------------------------------
# 2. ESTADÍSTICAS DESCRIPTIVAS
# ---------------------------------------------------------
media = np.mean(zanahorias)          # Promedio
mediana = np.median(zanahorias)      # Mediana (valor central)
moda = st.mode(zanahorias)          # Moda (valor más repetido)
desviacion = np.std(zanahorias)      # Desviación estándar

# ---------------------------------------------------------
# 3. MOSTRAR RESULTADOS EN CONSOLA
# ---------------------------------------------------------
print("Analisis del concurso de zanahorias")
print(f"Media: {media} gramos")
print(f"Mediana: {mediana} gramos")
print(f"Moda: {moda} gramos")
print(f"Desviacion standard: {desviacion} gramos")

# ---------------------------------------------------------
# 4. EXTRA 1: VISUALIZACIÓN CON MATPLOTLIB
# ---------------------------------------------------------
plt.hist(zanahorias, bins=6, color='orange', edgecolor='black')
plt.axvline(media, color='red', linestyle='dashed', linewidth=2, label=f'Media: {media:.1f}g')
plt.title('Distribución de pesos de las zanahorias concursantes')
plt.xlabel('Peso (gramos)')
plt.ylabel('Cantidad de zanahorias')
plt.legend()
plt.show()

# ---------------------------------------------------------
# 5. EXTRA 2: INGRESO INTERACTIVO DE NUEVOS PESOS
# ---------------------------------------------------------
print("\n--- Agregar nuevas zanahorias ---")
entrada = input("Ingresa los pesos de las nuevas zanahorias separados por comas: ")

if entrada.strip():
    # Convertimos la entrada de texto en una lista de números
    nuevas_zanahorias = [float(peso.strip()) for peso in entrada.split(",")]

    # Combinamos la lista original con las nuevas entradas
    todas_las_zanahorias = zanahorias + nuevas_zanahorias

    # Recalculamos las estadísticas generales
    nueva_media = np.mean(todas_las_zanahorias)
    nueva_desviacion = np.std(todas_las_zanahorias)

    print(f"\nTotal de zanahorias analizadas: {len(todas_las_zanahorias)}")
    print(f"Nueva media: {nueva_media:.2f} gramos")
    print(f"Nueva desviacion standard: {nueva_desviacion:.2f} gramos")
```
