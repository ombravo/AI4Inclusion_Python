```python
# Importamos la librería pandas para manejar estructuras de datos tabulares (DataFrames)
import pandas as pd
# Importamos 'List' del módulo typing para definir los tipos de datos esperados en los argumentos
from typing import List

# Definimos una función que toma una lista de listas de enteros y devuelve un DataFrame de pandas
def createDataframe(student_data: List[List[int]]) -> pd.DataFrame:
    # Definimos los nombres que tendrán las columnas de nuestra tabla
    column_names = ["student_id", "age"]

    # Creamos el DataFrame utilizando los datos de los estudiantes y las columnas especificadas
    result_dataframe = pd.DataFrame(student_data, columns=column_names)

    # Retornamos el DataFrame creado (esto es lo que evalúa LeetCode)
    return result_dataframe

# Datos de ejemplo: Una lista que contiene sublistas con el formato [id_estudiante, edad]
datos_estudiantes = [
    [1, 15],
    [2, 11],
    [3, 11],
    [4, 20]
]

# Llamamos a la función createDataframe pasándole los datos de prueba
df_resultado = createDataframe(datos_estudiantes)

# Usamos print() en lugar de display() para evitar errores en entornos estándar como LeetCode
print(df_resultado)
```
