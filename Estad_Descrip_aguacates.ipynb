```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def obtener_datos():
    """Permite al usuario ingresar datos manualmente o usar datos precargados."""
    print("--- CONCURSO DE AGUACATES GIGANTES 🥑 ---")
    print("1. Usar conjunto de datos precargados")
    print("2. Ingresar pesos manualmente")

    opcion = input("Seleccione una opción (1 o 2): ").strip()

    if opcion == "2":
        print("\nIngrese los pesos en gramos separados por comas (ej. 450, 520, 610, 480):")
        entrada = input("> ")
        pesos = [float(p.strip()) for p in entrada.split(",") if p.strip()]
    else:
        # Conjunto de datos precargados de muestra (en gramos)
        pesos = [
            420, 550, 580, 610, 490, 530, 470, 510, 500, 520,
            780, 850, 310, 480, 500, 530, 590, 620, 460, 540,
            920, 350, 510, 500, 480, 560, 520, 530, 600, 450
        ]
        print("\nSe han cargado 30 muestras de prueba.")

    return pd.Series(pesos, name="Peso_g")

def definir_rango_ideal():
    """Permite al usuario definir dinámicamente el rango ideal de peso."""
    try:
        min_i = float(input("\nDefina el peso MÍNIMO ideal (g) [por defecto 400]: ") or 400)
        max_i = float(input("Defina el peso MÁXIMO ideal (g) [por defecto 600]: ") or 600)
        return min_i, max_i
    except ValueError:
        print("Entrada no válida. Usando rango predeterminado (400g - 600g).")
        return 400.0, 600.0

def calcular_estadisticas(serie, min_ideal, max_ideal):
    """Calcula todas las métricas solicitadas (básicas y avanzadas)."""
    media = serie.mean()
    mediana = serie.median()

    # Moda (puede haber más de una)
    modas = serie.mode()
    moda_str = ", ".join(map(str, modas.tolist()))

    std = serie.std()
    cv = (std / media) * 100 if media != 0 else 0
    max_peso = serie.max()
    min_peso = serie.min()

    # Conteo en rango ideal
    en_rango = serie[(serie >= min_ideal) & (serie <= max_ideal)].count()
    pct_en_rango = (en_rango / len(serie)) * 100

    # Extra avanzado: Asimetría (Skewness)
    asimetria = serie.skew()

    res = {
        "Cantidad total": len(serie),
        "Media (promedio)": round(media, 2),
        "Mediana": round(mediana, 2),
        "Moda": moda_str,
        "Desviación Estándar": round(std, 2),
        "Coeficiente de Variación (%)": round(cv, 2),
        "Aguacate más pesado": max_peso,
        "Aguacate más ligero": min_peso,
        "En rango ideal": f"{en_rango} ({pct_en_rango:.1f}%)",
        "Asimetría (Skewness)": round(asimetria, 2)
    }
    return res

def generar_visualizaciones(serie, min_ideal, max_ideal):
    """Genera y muestra el histograma y el diagrama de caja."""
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))

    # Histograma
    axes[0].hist(serie, bins=8, color='#2e7d32', edgecolor='black', alpha=0.75)
    axes[0].axvline(min_ideal, color='red', linestyle='--', linewidth=2, label=f'Mín ideal ({min_ideal}g)')
    axes[0].axvline(max_ideal, color='orange', linestyle='--', linewidth=2, label=f'Máx ideal ({max_ideal}g)')
    axes[0].set_title('Distribución de Pesos de Aguacates')
    axes[0].set_xlabel('Peso (g)')
    axes[0].set_ylabel('Frecuencia')
    axes[0].legend()
    axes[0].grid(True, linestyle=':', alpha=0.6)

    # Boxplot (Valores atípicos)
    axes[1].boxplot(serie, vert=True, patch_artist=True,
                    boxprops=dict(facecolor='#aed581', color='#1b5e20', linewidth=1.5),
                    medianprops=dict(color='darkred', linewidth=2),
                    flierprops=dict(marker='o', markerfacecolor='red', markersize=8))
    axes[1].set_title('Diagrama de Caja (Boxplot) y Outliers')
    axes[1].set_ylabel('Peso (g)')
    axes[1].grid(True, linestyle=':', alpha=0.6)

    plt.tight_layout()
    plt.show()

def guardar_reporte(serie, estadisticas):
    """Guarda los resultados en archivos .csv y .txt."""
    # Guardar resumen en TXT
    with open("reporte_aguacates.txt", "w", encoding="utf-8") as f:
        f.write("=== REPORTE DEL CONCURSO DE AGUACATES GIGANTES ===\n\n")
        for clave, valor in estadisticas.items():
            f.write(f"{clave}: {valor}\n")

    # Guardar datos estructurados en CSV
    df_export = pd.DataFrame({
        "ID_Aguacate": range(1, len(serie) + 1),
        "Peso_g": serie.values
    })
    df_export.to_csv("datos_aguacates.csv", index=False)
    print("\n[✔] Reporte guardado con éxito en 'reporte_aguacates.txt' y 'datos_aguacates.csv'.")

# --- EJECUCIÓN PRINCIPAL ---
if __name__ == "__main__":
    datos = obtener_datos()
    min_i, max_i = definir_rango_ideal()
    stats = calcular_estadisticas(datos, min_i, max_i)

    print("\n" + "="*45)
    print("      RESULTADOS DEL ANÁLISIS ESTADÍSTICO")
    print("="*45)
    for k, v in stats.items():
        print(f" • {k}: {v}")
    print("="*45)

    generar_visualizaciones(datos, min_i, max_i)
    guardar_reporte(datos, stats)
```

pandas: Permite estructurar los datos en series y dataframes, facilitando el cálculo estadístico rápido y la exportación de datos.

Pregunta interactivamente al usuario los límites de peso mínimo y máximo que se consideran "ideales" para la competencia.

Media (mean()) y Mediana (median()).
Moda (mode()): Encuentra el peso o pesos que más se repiten.
Desviación Estándar (std()) y Coeficiente de Variación (la desviación estándar dividida por la media, expresada en porcentaje) para medir la dispersión.
Valores Extremos: El aguacate más pesado (max()) y el más ligero (min()).
Conteo en Rango Ideal: Filtra los datos que se encuentran dentro del rango definido previamente y calcula qué porcentaje representan sobre el total.
Asimetría (Skewness) (skew()): Indica hacia qué lado se sesga la distribución de los datos.

Genera una ventana gráfica dividida en dos subgráficos (figura de 1x2):

Histograma (izquierda): Muestra la distribución de frecuencias de los pesos en barras verdes. Incluye dos líneas verticales punteadas que representan los límites del rango ideal (mínimo y máximo).
Diagrama de Caja / Boxplot (derecha): Sirve para visualizar la dispersión, la mediana, los cuartiles y detectar de forma automática valores atípicos (outliers), los cuales se representan como puntos rojos fuera de los bigotes del gráfico.

Se encarga de la persistencia de datos guardando los resultados en dos archivos locales:

reporte_aguacates.txt: Un reporte de texto plano donde se escribe de forma estructurada cada una de las métricas estadísticas calculadas.
datos_aguacates.csv: Un archivo separado por comas con una tabla estructurada que contiene los identificadores (ID_Aguacate) y los pesos correspondientes, útil para abrirlo en Excel o herramientas similares.

```python

import pandas as pd
import matplotlib.pyplot as plt

# Datos precargados
pesos_iniciales = [
    350, 420, 480, 500, 550, 600, 420, 580, 390, 450,
    520, 610, 470, 490, 530, 420, 700, 380, 560, 950
]

df = pd.DataFrame({"peso": pesos_iniciales})
min_ideal, max_ideal = 400, 600


def mostrar_estadisticas():
    pesos = df["peso"]

    print("\n--- ESTADÍSTICAS DESCRIPTIVAS ---")
    print(f"Media: {pesos.mean():.2f} g")
    print(f"Mediana: {pesos.median():.2f} g")
    print(f"Moda: {pesos.mode().tolist()} g")
    print(f"Desviación estándar: {pesos.std():.2f} g")
    print(f"Coeficiente de variación: {pesos.std() / pesos.mean() * 100:.2f}%")
    print(f"Asimetría: {pesos.skew():.2f}")

    print("\n--- PESOS EXTREMOS ---")
    print(f"Más pesado: {pesos.max()} g")
    print(f"Más ligero: {pesos.min()} g")

    ideales = pesos.between(min_ideal, max_ideal)
    print(f"\n--- RANGO IDEAL ({min_ideal:g} - {max_ideal:g} g) ---")
    print(f"Aguacates dentro del rango: {ideales.sum()}")
    print(f"Porcentaje: {ideales.mean() * 100:.2f}%")


def mostrar_graficos():
    fig, axes = plt.subplots(1, 2, figsize=(11, 4))

    axes[0].hist(df["peso"], bins=8, edgecolor="black")
    axes[0].set_title("Distribución de pesos")
    axes[0].set_xlabel("Peso (g)")
    axes[0].set_ylabel("Cantidad de aguacates")

    axes[1].boxplot(df["peso"])
    axes[1].set_title("Valores atípicos")
    axes[1].set_ylabel("Peso (g)")

    plt.tight_layout()
    plt.show()


def ingresar_datos():
    global df

    try:
        entrada = input("Introduce los pesos separados por comas: ")
        pesos = [float(p.strip()) for p in entrada.split(",")]

        if len(pesos) < 3 or any(p <= 0 for p in pesos):
            print("Introduce al menos 3 pesos positivos.")
            return

        df = pd.DataFrame({"peso": pesos})
        print("Datos actualizados correctamente.")

    except ValueError:
        print("Error: introduce únicamente números separados por comas.")


def definir_rango():
    global min_ideal, max_ideal

    try:
        minimo = float(input("Peso mínimo ideal (g): "))
        maximo = float(input("Peso máximo ideal (g): "))

        if minimo <= 0 or maximo < minimo:
            print("Rango inválido.")
            return

        min_ideal, max_ideal = minimo, maximo
        print("Rango actualizado correctamente.")

    except ValueError:
        print("Debes introducir valores numéricos.")


def exportar_analisis():
    pesos = df["peso"]
    ideales = pesos.between(min_ideal, max_ideal)

    with open("analisis_aguacates.txt", "w", encoding="utf-8") as archivo:
        archivo.write("CONCURSO DE AGUACATES GIGANTES\n\n")
        archivo.write(f"Cantidad: {len(pesos)}\n")
        archivo.write(f"Media: {pesos.mean():.2f} g\n")
        archivo.write(f"Mediana: {pesos.median():.2f} g\n")
        archivo.write(f"Moda: {pesos.mode().tolist()} g\n")
        archivo.write(f"Desviación estándar: {pesos.std():.2f} g\n")
        archivo.write(f"Coeficiente de variación: {pesos.std() / pesos.mean() * 100:.2f}%\n")
        archivo.write(f"Asimetría: {pesos.skew():.2f}\n")
        archivo.write(f"Más pesado: {pesos.max()} g\n")
        archivo.write(f"Más ligero: {pesos.min()} g\n")
        archivo.write(f"Rango ideal: {min_ideal:g} - {max_ideal:g} g\n")
        archivo.write(f"Cantidad en rango: {ideales.sum()}\n")
        archivo.write(f"Porcentaje en rango: {ideales.mean() * 100:.2f}%\n")

    print("Análisis guardado en analisis_aguacates.txt")

def exportar_csv():
    df.to_csv("aguacates.csv", index=False, encoding="utf-8-sig")
    print("Datos exportados correctamente a aguacates.csv")

# Menú principal
while True:
    print("\n===== CONCURSO DE AGUACATES GIGANTES =====")
    print("1. Usar datos precargados")
    print("2. Introducir pesos manualmente")
    print("3. Definir rango ideal")
    print("4. Mostrar estadísticas y análisis")
    print("5. Mostrar gráficos")
    print("6. Exportar análisis a TXT")
    print("7. Exportar análisis a CSV")
    print("0. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        df = pd.DataFrame({"peso": pesos_iniciales})
        print("Datos precargados seleccionados.")

    elif opcion == "2":
        ingresar_datos()

    elif opcion == "3":
        definir_rango()

    elif opcion == "4":
        mostrar_estadisticas()

    elif opcion == "5":
        mostrar_graficos()

    elif opcion == "6":
        exportar_analisis()

    elif opcion == "0":
        print("¡Hasta pronto!")
        break
    elif opcion == "7":
      exportar_csv()

    else:
        print("Opción no válida. Inténtalo de nuevo.")
```
